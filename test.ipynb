```python
#https://t.me/socks?server=209.200.239.99&port=59101&user=hiostad6&pass=4GwW5dAw9E

file = open("proxy.txt","r")

lines = file.readlines()
sockss = []
for line in lines:
    line = line.strip()
    ip_port = line.split(":")
    ip = ip_port[0]
    port = int(ip_port[1])
    username = ip_port[2]
    password = ip_port[3]
    sockss.append(f"https://t.me/socks?server={ip}&port={port}&user={username}&pass={password}\n")

with open("proxy_links.txt","w") as f:
    f.writelines(sockss)
```
